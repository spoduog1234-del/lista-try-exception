```python
#Maior entre "três" números .6
lisnum=[]
while True:
  try:
    x=int(input("Insira quantos números quer: "))
    if x<=0:
      raise ValueError()
    break
  except ValueError:
    print("Incorreto")
for i in range(x):
  while True:
    try:
      a=int(input("Insira um número: "))
      lisnum.append(a)
      break
    except ValueError:
      print("Incorreto")
lisnum.sort()
print("O Maior número é:",lisnum[-1])
```

```python
#Contagem crescente .7
while True:
  try:
    N=int(input("Insira um número N: "))
    if N<=0:
      raise ValueError()
    break
  except ValueError:
    print("Expressão inválida (texto ou número menor ou igual á 0.)")
for i in range(N):
  print(i+1)
```

```python
#Soma de 1 até N .8
som=0
while True:
  try:
    N2=int(input("Insira um número N: "))
    if N2<=0:
      raise ValueError()
    break
  except ValueError:
    print("Expressão inválida (texto ou número menor ou igual á 0.)")
for i in range(N2):
  som+=i+1
print("Soma =",som)
```

```python
#Contando números pares. 9
parcount=0
for i in range(10):
  while True:
    try:
      numb=int(input("Insira um número: "))
      if numb%2==0:
        parcount+=1
      break
    except ValueError:
      print("Incorreto, insira apenas números(inteiros).")
print("São",parcount,"números pares.")
```

```python
#Soma apenas dos positivos. 10
somapos=0
for i in range(8):
  while True:
    try:
      numb2=int(input("Insira um número: "))
      if numb2>=0:
        somapos+=numb2
      break
    except ValueError:
      print("Incorreto, insira apenas números(inteiros).")
print("Soma dos positivos:",somapos)
```

```python
#Maior valor digitado .11
maior=-float("inf")
valores=[]
for i in range(6):
  while True:
    try:
      numb3=int(input("Insira um número: "))
      valores.append(numb3)
      if maior<numb3:
        maior=numb3
      break
    except ValueError:
      print("Incorreto, insira apenas números(inteiros).")
print("Os Valores são:",valores,"\nO Maior é:",maior)
```

```python
#Tabuada .12
while True:
  try:
    tabx=int(input("Insira o número gerador da tabuada: "))
    break
  except ValueError:
    print("Incorreto, insira apenas números(inteiros).")
print("Tabuada do",str(tabx)+"!")
for i in range(10):
  print(str(i+1)+"x"+str(tabx),"=",str((i+1)*tabx))
```

```python
#Tentativas de senha .13
senha=1234
inputsenha=0
while senha!=inputsenha:
  try:
    inputsenha=int(input("Insira sua senha: "))
    if inputsenha!=senha:
      print("Senha incorreta.\n")
  except ValueError:
    print("Senha de padrão incorreto.\no padrão é: Senha númerica e inteira.\n")
print("Senha correta!")
```

```python
#Número secreto .14
s=7
tent=0
while True:
  try:
    tent=int(input("Insira um número: "))
    if s>tent:
      print("Maior")
    elif s<tent:
      print("Menor")
    elif s==7:
      print("Correto!")
      break
  except ValueError:
    print("Entrada Inválida, apenas números inteiros.")
```

```python
#Analisando uma sequência .15
sq=[]
quant=neg=pos=soma=0
while True:
  try:
    add=int(input("Adicione um número: "))
    if add==0:
      break
    sq.append(add)
    quant+=1
    soma+=add
    if add>0:
      pos+=1
    else:
      neg+=1
  except ValueError:
    print("Entrada Inválida.")
print("Quantidade de números:",quant,"\nSoma:",soma,"\nPositivos:",pos,"\nNegativos:",neg)
```
